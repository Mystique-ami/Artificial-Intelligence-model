```python
from keras._tf_keras.keras.preprocessing.image import ImageDataGenerator
```

```python
train_datagen =ImageDataGenerator(rescale = 1./255,shear_range= 0.2,zoom_range=0.2)
```

```python
training_set1 = train_datagen.flow_from_directory('/content/ASL/American Sign Language (ASL)/data', target_size=(64,64), batch_size=32, class_mode='categorical')
```

```python
test_datagen1 = ImageDataGenerator(rescale=1./255)
```

```python
test_set1 = test_datagen1.flow_from_directory('/content/ASL/American Sign Language (ASL)/data', target_size=(64,64), batch_size=32, class_mode='categorical')
```

```python
cnn1 = tf.keras.models.Sequential()
```

```python
cnn1.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu', input_shape=[64,64,3]))
```

```python
cnn1.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

```python
cnn1.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
```

```python
cnn1.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

```python
cnn1.add(tf.keras.layers.Flatten())
```

```python
cnn1.add(tf.keras.layers.Dense(units=120, activation='relu'))
```

```python
cnn1.add(tf.keras.layers.Dense(units=27, activation='softmax'))
```

```python
cnn1.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
cnn1.fit(x=training_set1, validation_data=test_set1, epochs=30)
early_stop = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
history = cnn1.fit(x=training_set1, validation_data=test_set1, epochs=30, callbacks=[early_stop])
```

```python
import numpy as np
```

```python
from keras.preprocessing import image
```

```python
test_image1 = image.load_img('/content/ASL/American Sign Language (ASL)/dataAT/A/1.jpg', target_size=(64,64))
```

```python
test_image1 = image.img_to_array(test_image1)
```

```python
test_image1 = test_image1/255.0
```

```python
test_image1 = np.expand_dims(test_image1, axis=0)
```

```python
results = cnn1.predict(test_image1)
```

```python
class_labels = training_set1.class_indices
reversed_class_labels = dict((v, k) for k, v in class_labels.items())

predicted_class_index = np.argmax(results[0])
prediction = reversed_class_labels[predicted_class_index]
```

```python
confidence = results[0][predicted_class_index] * 100

print("Predicted class index:", predicted_class_index)
print("Predicted letter:", prediction)
print("Confidence:", round(confidence, 2), "%")
```
